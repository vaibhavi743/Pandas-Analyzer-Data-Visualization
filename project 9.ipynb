```python
df1 = df.iloc[:5]
df2 = df.iloc[5:10]

combined_df = pd.concat([df1, df2], ignore_index=True)

print("Combined DataFrame:")
display(combined_df)
```

```python
pivot_table = pd.pivot_table(
    df,
    index="Product",
    columns="Year",
    values="Sales",
    aggfunc="sum"
)

print("Pivot Table:")
display(pivot_table)
```

```python
reindexed_df=df.set_index("SalesID")

print("dataframe after re-indexing:")
display(reindexed_df)
```

```python
grouped_data = df.groupby("Region")["Sales"].mean()

print("Average Sales by Region:")
display(grouped_data)
```

```python
df["Region_Average_Sales"] = df.groupby("Region")["Sales"].transform("mean")

print("GroupBy + Transform:")
display(df[["Region", "Sales", "Region_Average_Sales"]])
```

```python
df.isnull().sum()
```

```python
missing_rows = df[df.isnull().any(axis=1)]

if missing_rows.empty:
    print("No missing values found in the dataset!")
else:
    display(missing_rows)
```

```python
df_mean = df.copy()

numeric_columns = df_mean.select_dtypes(include="number").columns

for column in numeric_columns:
    df_mean[column] = df_mean[column].fillna(
        df_mean[column].mean()
    )

print("Missing numeric values filled with mean:")
display(df_mean)
```

```python
df_drop=df.dropna()

print("Data after dropping missing values:")
display(df_drop)
```

```python
print("Mean Sales:", df["Sales"].mean())
print("Total Sales:", df["Sales"].sum())
print("Number of Sales Records:", df["Sales"].count())
```

```python
sales_by_product = df.groupby("Product")["Sales"].sum()

plt.figure(figsize=(8, 5))
plt.bar(sales_by_product.index, sales_by_product.values)

plt.title("Total Sales by Product")
plt.xlabel("Product")
plt.ylabel("Total Sales")

plt.show()
```

```python
year_sales = df.groupby("Year")["Sales"].sum()

plt.figure(figsize=(8, 5))
plt.plot(year_sales.index, year_sales.values, marker="o")

plt.title("Sales Trend by Year")
plt.xlabel("Year")
plt.ylabel("Total Sales")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.scatter(df["Sales"], df["Year"])

plt.title("Sales vs Year")
plt.xlabel("Sales")
plt.ylabel("Year")

plt.show()
```

```python
region_sales = df.groupby("Region")["Sales"].sum()

plt.figure(figsize=(7, 7))

plt.pie(
    region_sales.values,
    labels=region_sales.index,
    autopct="%1.1f%%"
)

plt.title("Sales Distribution by Region")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.hist(df["Sales"], bins=8, edgecolor="black")

plt.title("Distribution of Sales")
plt.xlabel("Sales")
plt.ylabel("Frequency")

plt.show()
```

```python
stack_data = df.groupby(["Year", "Region"])["Sales"].sum().unstack(fill_value=0)

plt.figure(figsize=(9, 5))

plt.stackplot(
    stack_data.index,
    *[stack_data[region] for region in stack_data.columns],
    labels=stack_data.columns
)

plt.title("Sales by Year and Region")
plt.xlabel("Year")
plt.ylabel("Sales")
plt.legend()

plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.scatter(df["Sales"], df["Year"])

plt.title("Sales vs Year")
plt.xlabel("Sales")
plt.ylabel("Year")

plt.savefig("scatter_plot.png", dpi=300, bbox_inches="tight")

plt.show()

print("Visualization saved successfully as scatter_plot.png")
```
